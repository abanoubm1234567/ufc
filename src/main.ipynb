```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/ufc-master.csv')
```

```python
cumulative = 0
equal_cumulative = 0
wins = 0
i = 0
df = df.dropna(subset=['RedOdds', 'BlueOdds', 'RedExpectedValue', 'BlueExpectedValue'])
x_values = []
y_values = []
equal_y_values = []
for row in df.itertuples():
 i+=1
 x_values.append(i)
 if (row.RedOdds < row.BlueOdds):
  if (row.Winner == 'Red'):
   cumulative += row.RedExpectedValue
   wins += 1
   
  else:
   cumulative -= 100
   
 else:
  if (row.Winner == 'Blue'):
   cumulative += row.BlueExpectedValue
   wins += 1
  else:
   cumulative -= 100
 equal_cumulative = wins/i;
 y_values.append(cumulative)
 equal_y_values.append(equal_cumulative)
```

```python
plt.plot(x_values, y_values)
plt.xlabel('Fight #')
plt.ylabel('Cumulative Earnings')
plt.title('Gambling Earnings When Strictly Choosing Favorite')
plt.show()
```

```python
plt.plot(x_values, equal_y_values)
plt.xlabel('Fight #')
plt.ylabel('Cumulative Correct v. Incorrect Predictions')
plt.title('Cumulative Fight Prediction Score When Strictly Choosing Favorite')
plt.show()
```

```python
print('The favorite wins about',equal_cumulative,"percent of the time.")
```

```python
#What about betting on the underdog?

cumulative = 0
equal_cumulative = 0
y_values = []
equal_y_values = []

for row in df.itertuples():
 if (row.RedOdds < row.BlueOdds):
  if (row.Winner == 'Blue'):
   cumulative += row.BlueExpectedValue
   equal_cumulative += 1
  else:
   cumulative -= 100
   equal_cumulative-=1
 else:
  if (row.Winner == 'Red'):
   cumulative += row.RedExpectedValue
   equal_cumulative += 1
  else:
   cumulative -= 100
   equal_cumulative-=1
 y_values.append(cumulative)
 equal_y_values.append(equal_cumulative)
```

```python
plt.plot(x_values, y_values)
plt.xlabel('Fight #')
plt.ylabel('Cumulative Earnings')
plt.title('Gambling Earnings When Strictly Choosing Underdog')
plt.show()
```

```python
plt.plot(x_values, equal_y_values)
plt.xlabel('Fight #')
plt.ylabel('Cumulative Correct v. Incorrect Predictions')
plt.title('Fight Prediction Score When Strictly Choosing Underdog')
plt.show()
```
