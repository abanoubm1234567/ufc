```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/ufc-master.csv')
```

```python
cumulative = 0
i = 0
df = df.dropna(subset=['RedOdds', 'BlueOdds', 'RedExpectedValue', 'BlueExpectedValue'])
x_values = []
y_values = []
for row in df.itertuples():
 i+=1
 x_values.append(i)

 if (row.RedOdds < row.BlueOdds):
  if (row.Winner == 'Red'):
   cumulative += row.RedExpectedValue
   #print('win',row.RedExpectedValue, ' - Cumulative:',cumulative)
  else:
   
   cumulative -= 100
   #print('loss -100', ' - Cumulative:',cumulative)
 else:
  if (row.Winner == 'Blue'):
   cumulative += row.BlueExpectedValue
   #print('win',row.BlueExpectedValue, ' - Cumulative:',cumulative)
   
  else:
   
   cumulative -= 100
   #print('loss -100', ' - Cumulative:',cumulative)
 y_values.append(cumulative)
```

```python
plt.plot(x_values, y_values)
```
